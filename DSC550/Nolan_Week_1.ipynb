```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/Users/doug/DSC 550/Video_Games_Sales_as_at_22_Dec_2016.csv")

print(df.head(10))
```

```python
print(df.shape)
# there are 16,719 rows and 16 columns in this data
# there are 16.7K video games and we have 16 columns of data on these games
```

```python
# find top 5 games by critic score

df_top_critic = df.sort_values(by= 'Critic_Score', ascending=False)[['Name', 'Platform', 'Critic_Score', 'Global_Sales']]
display(df_top_critic.head(5))
```

```python
# find the count of games in each genre
genre_count = df['Genre'].value_counts()

print('Games by Genre:')
display(genre_count)
```

```python
# find first 5 games in the df on the SNES
snes_games = df[df['Platform'] == 'SNES']
display(snes_games.head(5))
```

```python
# find 5 publisher with the highest global sales - sum function
pub_sales = (
    df.groupby('Publisher')['Global_Sales']
    .sum()
    .sort_values(ascending=False)
)

print('Top 5 Publishers by Total Globabl Sales:')
display(pub_sales.head(5))
```

```python
# create new column - % of global sales from North America. display first 5 rows of new df
df['NA_Sales_Pct'] = (
    df['NA_Sales'] / df['Global_Sales']) *100

display(
    df[
        ['Name',
         'NA_Sales',
         'Global_Sales',
         'NA_Sales_Pct']
        ].head()
)
```

```python
# find the count of null entries in each column
missing_values = df.isna().sum()

print('Missing Values in each column:')
display(missing_values)
```

```python
# caclulate median user score - will get error. find and rplace string with nan and then calculate median. then replace all null with median value
print(df['User_Score'].unique()[:40])
# found a nan and tbd value in here. nan is null - tbd is to be determined . weird rating but whatever

# replace tbd with nan
df['User_score'] = df['User_Score'].replace('tbd', np.nan)

#convert to numeric values and not string
df['User_Score'] = pd.to_numeric(df['User_Score'], errors='coerce')

```

```python
# find median score
median_user_score = df['User_Score'].median()
print(f'Median User Score: {median_user_score}')

# fill missing values with median
df['User_Score'] = df['User_Score'].fillna(median_user_score)

# validate
print('\nMissing User Scores Remaining:')
print(df['User_Score'].isna().sum())

display(df[['Name', 'User_Score']].head())
```
