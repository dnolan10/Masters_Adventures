# Week 1, Exercise 1

We want to look at data for the Facebook, Apple, Amazon, Netflix, and Google (FAANG) stocks, but we were given each as a separate CSV file (obtained using the stock_analysis package we will build in Chapter 7, Financial Analysis – Bitcoin and the Stock Market). Combine them into a single file and store the dataframe of the FAANG data as faang for the rest of the exercises:

Read in the aapl.csv, amzn.csv, fb.csv, goog.csv, and nflx.csv files.
Add a column to each dataframe, called ticker, indicating the ticker symbol it is for (Apple's is AAPL, for example); this is how you look up a stock. In this case, the filenames happen to be the ticker symbols.
Append them together into a single dataframe.
Save the result in a CSV file called faang.csv

```python
import pandas as pd
```

```python
# 1a. Read in the aapl.csv, amzn.csv, fb.csv, goog.csv, and nflx.csv files.
aapl = pd.read_csv("Bellevue/dsc530/Hands-On-Data-Analysis-with-Pandas-2nd-edition-master/ch_07/data/apple.csv")
amzn = pd.read_csv("Bellevue/dsc530/Hands-On-Data-Analysis-with-Pandas-2nd-edition-master/ch_07/data/amazon.csv")
fb = pd.read_csv("Bellevue/dsc530/Hands-On-Data-Analysis-with-Pandas-2nd-edition-master/ch_07/data/facebook.csv")
goog = pd.read_csv("Bellevue/dsc530/Hands-On-Data-Analysis-with-Pandas-2nd-edition-master/ch_07/data/google.csv")
nflx = pd.read_csv("Bellevue/dsc530/Hands-On-Data-Analysis-with-Pandas-2nd-edition-master/ch_07/data/netflix_jan_2021.csv")
```

```python
# 1b. Add a column to each dataframe, called ticker, indicating the ticker symbol it is for (Apple's is AAPL, for example);
# this is how you look up a stock. In this case, the filenames happen to be the ticker symbols.
aapl['ticker'] = 'AAPL'
amzn['ticker'] = 'AMZN'
fb['ticker'] = 'FB'
goog['ticker'] = 'GOOG'
nflx['ticker'] = 'NFLX'
```

```python
# 1c. Append them together into a single dataframe.
faang = pd.concat([aapl, amzn, fb, goog, nflx], ignore_index=True)
```

```python
# 1d. d) Save the result in a CSV file called faang.csv.
faang.to_csv('faang.csv', index=False)
faang.head()
```

# Exercise 2

With faang, use type conversion to cast the values of the date column into datetimes and the volume column into integers. Then, sort by date and ticker.

```python
# make sure date and volumes are expected data types and then sort by date.
faang['date'] = pd.to_datetime(faang['date'])
faang['volume'] = faang['volume'].astype(int)

faang = faang.sort_values(by=['date', 'ticker'])
faang.head()
```

#### do these things
1. Read in the covid19_cases.csv file. 
2. Create a date column using the data in the dateRep column and the pd.to_datetime() function.
3. Set the date column as the index and sort the index.
4. Replace all occurrences of United_States_of_America and United_Kingdom with USA and UK, respectively. Hint: the replace() method can be run on the dataframe as a whole.
5. Using the countriesAndTerritories column, filter the cleaned COVID-19 cases data down to Argentina, Brazil, China, Colombia, India, Italy, Mexico, Peru, Russia, Spain, Turkey, the UK, and the USA.
6. Pivot the data so that the index contains the dates, the columns contain the country names, and the values are the case counts (the cases column). Be sure to fill in NaN values with 0.

```python
#read pesky excel file, not CSV
covid = pd.read_excel('Bellevue/dsc530/COVID-19-geographic-disbtribution-worldwide-2020-12-14.xlsx')

covid.head()
```

```python
# ensure date datatype is compatible for sorting
covid['date'] = pd.to_datetime(covid['dateRep'], dayfirst=True)

covid = covid.set_index('date').sort_index()

covid = covid.replace({
    'United_States_of_America': 'USA',
    'United_Kingdom': 'UK'})

covid.head()
```

```python
#drill down to specific countries only
countries = [
    'Argentina', 'Brazil', 'China', 'Columbia', 'India', 'Italy', 'Mexico', 'Peru',
    'Russia', 'Spain', 'Turkey', 'UK', 'USA']

covid_filtered = covid[covid['countriesAndTerritories'].isin(countries)]
covid_filtered.head()
```

```python
#pivot with only the selected countries
covid_pivot = covid_filtered.pivot_table(
    index='date',
    columns='countriesAndTerritories',
    values='cases',
    fill_value=0)

covid_pivot.head()
```
