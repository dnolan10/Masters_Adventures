```python
import random
import pandas
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
```

# Exercise 5


### Using the data from exercise 4, calculate the following statistics without importing anything from the statistics module in the Python standard library, and then confirm your results match up to those that are obtained when using the statistics module (where possible):

```python
# mean calculation
n = len(salaries)
mean = sum(salaries) / n

print(mean)

# median calculation
sorted_data = sorted(salaries)
if n % 2 == 0:
    median = (sorted_data[n//2-1] + sorted_data[n//2]) / 2
else:
    median = sorted_data[n//2]

print(median)

# mode
mode = max(set(salaries), key=salaries.count)
print(mode)

# sample variance
variance = sum((x - mean) ** 2 for x in salaries) / (n-1)
print(variance)

# sample standard deviation
std_dev = variance ** 0.5
print(std_dev)
```

```python
import statistics as stats

print(
    stats.mean(salaries),
    stats.median(salaries),
    stats.mode(salaries),
    stats.variance(salaries),
    stats.stdev(salaries)
)
```

# Exercise 6

```python
# range
data_range = max(salaries) - min(salaries)

# coefficient of variation
cv = stats.stdev(salaries) / stats.mean(salaries)

# quartiles
q1 = stats.quantiles(salaries, n=4)[0]
q3 = stats.quantiles(salaries, n=4)[2]

# interquartile range
iqr = q3 - q1

# quartile coefficient of dispersion
qcd = (q3 - q1) / (q3 + q1)

print(data_range, cv, iqr, qcd)
```
